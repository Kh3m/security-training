# Bits, Bytes and Hex
The first foundation for this repo. Before working with hashes, HMAC or signatures, it helps to know what a byte really is and the different ways we write bytes down.

## Step 1: A byte is 8 bits
A **bit** is a single 0 or 1. A **byte** is 8 bits in a row. Each bit doubles the number of possible patterns, so one byte can hold 2 × 2 × 2 × 2 × 2 × 2 × 2 × 2 = 256 different values. That's why a byte is always a number from **0 to 255**.

```python
# Step 1: A byte is 8 bits, so it holds a number from 0 to 255
print(2 ** 8, "possible values in one byte")
print(bytes([0, 104, 255]))
# bytes([256]) # uncomment to see the error: too big for one byte
```

When Python prints bytes, it uses a shortcut. If a byte’s number matches a normal printable ASCII character, it shows that character instead. **104** is the letter **h**, so Python shows h. The numbers **0** and **255** have no printable character, so Python shows them as **\x00** and **\xff**

```python
data = bytes([0, 104, 255])

print(data)          # b'\x00h\xff'  (Python's shortcut display)
print(len(data))     # 3             (still three bytes)
print(list(data))    # [0, 104, 255] (the real numbers)
print(data.hex())    # 0068ff        (every byte shown the same way: 00, 68, ff)

print(bytes([104, 105]))   # b'hi'   (both are printable, so no \x at all)
```

## Step 2: Writing a number as binary and hex
`format(value, code)` turns a number into text. The code tells Python how to write it. Read the code from the right:

| Code | Letter | Number | Leading 0 |
|------|--------|--------|-----------|
| `08b` | `b` = binary | at least 8 characters | fill the front with zeros |
| `02x` | `x` = hex | at least 2 characters | fill the front with zeros |

Capital `X` gives capital hex letters. f-strings use the same codes, for example `f"{n:08b}"`. The shortcuts `bin()` and `hex()` add a `0b` or `0x` label at the front but don't pad.

```python
# Step 2: Writing a number as binary and hex with format codes
# b = binary, x = hex, the number = minimum width, 0 = pad with zeros
n = 5
print(format(n, "b"), format(n, "08b"))
print(format(n, "x"), format(n, "02x"))
print(f"{n:08b}", f"{n:02x}", f"{n:02X}")   # same codes in f-strings
print(bin(n), hex(n))                       # shortcuts with 0b and 0x labels
```

### Why the padding matters
A byte is always 8 bits and always 2 hex characters. Without padding, the number 5 shows as `5` in hex, which looks like half a byte. When many bytes are joined together, like in a hash, you couldn't tell where one byte ends and the next starts. Padding keeps every byte the same width.

## Step 3: One hex character = 4 bits
Hex uses 16 symbols (`0` to `9`, then `a` to `f`), and 4 bits also give exactly 16 patterns. So each hex character stands for 4 bits, and a byte (8 bits) needs **2 hex characters**. The cell below splits each byte into two halves of 4 bits so you can see each half become one hex character.

```python
# Step 3: One hex character = 4 bits, so one byte = 2 hex characters
for n in [104, 255]:
    bits = format(n, "08b")
    print(n, bits[:4], bits[4:], "->", format(n, "02x"))
```

## Step 4: The rule, applied to SHA-256
Two simple rules come from the steps above:

- **bytes × 8 = bits**
- **bytes × 2 = hex characters**

Clearer:
- **1 byte = 8 bits**, so number of bits = number of bytes × 8
- **1 byte = 2 hex characters**, so number of hex characters = number of bytes × 2

SHA-256 gives 32 bytes. 32 × 8 = 256 bits, which is where the name comes from. 32 × 2 = 64, which is why a SHA-256 hex digest is always 64 characters long.

```python
# Step 4: The rule, applied to SHA-256
# bytes x 8 = bits, bytes x 2 = hex characters
import hashlib
d = hashlib.sha256(b"hello").digest()
print(len(d), "bytes =", len(d) * 8, "bits =", len(d.hex()), "hex characters")
```

## Summary
- A bit is 0 or 1. A byte is 8 bits and holds a number from 0 to 255.
- `format()` codes like `08b` and `02x` write numbers in binary or hex with a fixed width.
- One hex character is 4 bits, so one byte is 2 hex characters.
- bytes × 8 = bits, bytes × 2 = hex characters.
